```python
from langchain.chat_models import BaseChatModel
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
import os

load_dotenv()

def get_model() -> BaseChatModel:

    model = ChatGoogleGenerativeAI(
        model = os.getenv("GEMINI_MODEL_ID")
    )

    return model


llm = get_model()
```

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template(
    template = "What is the capital of {country}?"
)

llm = get_model()

chain = prompt | llm

response = chain.invoke({
    "country": "India"
})

print(response.text)
```

```python
from langchain_core.prompts import ChatPromptTemplate

llm = get_model()

prompt = ChatPromptTemplate([
    ("system", "You are a helpful {agent_type} agent"),
    ("user", "What is the best monthly {agent_type} plan in {country}?")
])

chain = prompt | llm 

# prompt.invoke({
#     "agent_type": "Financial",
#     "country": "India"
# })

response = chain.invoke({
    "agent_type": "financial",
    "country": "India"
})

print(response.text)

```

```python
from langchain_core.output_parsers import StrOutputParser

output = StrOutputParser()

chain = prompt | llm | output

response = chain.invoke({
    "agent_type": "Health Insurance",
    "country": "India"
})

print(response)
```

```python
from pydantic import BaseModel, Field


class OutputFormat(BaseModel):
    payment_id: str = Field(description="Payment ID")
    status: str = Field(description="Payment Status")
    notification: str = Field(description="Payment status Notification")


prompt_message = ChatPromptTemplate([
    ("system", "Generate the paymet details in a structured manner"),
    ("user", "Get me the payment details of the payment id {payment_id}"),
    ("ai", "This payment is in Pending since last 24 hours and the status is reported to the enduser via email and whatsapp"),
    ("user", "Received, Thankyou!")
])


chain = prompt_message | llm.with_structured_output(OutputFormat)

response = chain.invoke({
    "payment_id": "12345678"
})

print(response)
```

```python
# Loading and Chunking a sample text file using simple Character splitter method

from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import CharacterTextSplitter

doc_loader = TextLoader(
    file_path="./data/test_file.txt",
    autodetect_encoding=True
)

splitter = CharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=20
)

chunks = doc_loader.load_and_split(splitter)

print(chunks[0].page_content)
```

```python
# Loading and Chunking CSV file using Recursive Chunking Method

from langchain_community.document_loaders import CSVLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

csv_loader = CSVLoader(
    file_path="./data/test_file.csv",
)

splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap =0,
    separators=["\n"]
)

chunks = csv_loader.load_and_split(splitter)


print(chunks[0].page_content)
print(chunks[0].metadata)
chunks[0].metadata['source'].split("/")[-1].split(".")[0]
```

```python
# Embedding list of Documents

from langchain_core.documents import Document
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_chroma import Chroma
from dotenv import load_dotenv
import os

load_dotenv()


documents = [
    Document(
        page_content="This is a document for the Refund Process.",
        metadata = {"page_number": 1, "description": "Heading"}
    ),
    Document(
        page_content="Refund is a process of settling the amount back to the customer.",
        metadata={"page_number": 2, "description": "definition"}
    ),
    Document(
        page_content="A refund request must be raised within 30 days of purchase",
        metadata = {"page_number":3, "description": "Limitation"}
    )
]

# response = ChatGoogleGenerativeAI(
#     model= os.getenv("GEMINI_MODEL_ID")
# )

# print(response.invoke("What is the capital of France?"))

embedding_function = GoogleGenerativeAIEmbeddings(
    model = "gemini-embedding-001",
)

embedding = embedding_function.embed_query("What is the refund process?")

texts = [doc.page_content for doc in documents]

#print(docs)

results = embedding_function.embed_documents(texts)


```

```python
print(len(embedding))

print(embedding)

len(results)

for result in results:
    print(result)
```

```python
# Indexing into ChromaDB

chroma_db = Chroma(
    collection_name="sample-collection",
    embedding_function=embedding_function,
    persist_directory="./vector_databases/chromadb/"
)

indexing = chroma_db.add_documents(documents=documents)

print(indexing)
```

```python
# Indexing into FAISS

from langchain_community.vectorstores import FAISS
from langchain_community.docstore.in_memory import InMemoryDocstore

import faiss


index = faiss.IndexFlatL2(len(embedding_function.embed_query("hello world")))



faiss_db = FAISS(
    embedding_function=embedding_function,
    index = index,
    docstore= InMemoryDocstore(),
    index_to_docstore_id={}
)

faiss_indexing = faiss_db.add_documents(documents=documents)

print(faiss_indexing)

faiss_db.save_local("./vector_databases/FAISS/")
```

```python
# Querying Chroma DB

chroma_search_results = chroma_db.as_retriever().invoke("Refund?")

print(chroma_search_results)


for result in chroma_search_results:
    print(result.page_content)
```

```python
# Querying FAISS DB

faiss_search_results = faiss_db.as_retriever().invoke("Refund?")

print(faiss_search_results)

for result in faiss_search_results:
    print(result.page_content)
```
